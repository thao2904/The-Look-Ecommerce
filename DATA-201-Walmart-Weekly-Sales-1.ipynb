```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import seaborn as sns # data visualisation
import matplotlib.pyplot as plt # plot
import os
import warnings

#import funcs
from datetime import datetime
from matplotlib.pyplot import cm
```

```python
df_train = pd.read_csv('Downloads/train.csv', parse_dates=['Date']).set_index(keys=['Store','Dept','Date'], drop=False)
df_train
```

# EDA

```python
# Remove duplicate rows based on all columns
df_train.drop_duplicates()
```

```python
df_train.describe()
```

```python
df_train.dtypes
```

```python
# Convert 'Date' to datetime
df_train['Date'] = pd.to_datetime(df_train['Date'])

# Ensure 'Store' and 'Dept' are integers
df_train['Store'] = df_train['Store'].astype(int)
df_train['Dept'] = df_train['Dept'].astype(int)

# Ensure 'Weekly_Sales' is float
df_train['Weekly_Sales'] = df_train['Weekly_Sales'].astype(float)

# Convert 'IsHoliday' to boolean
df_train['IsHoliday'] = df_train['IsHoliday'].map({'TRUE': True, 'FALSE': False}).astype(bool)

# Verify data types after conversion
print("\nData Types After Conversion:")
print(df_train.dtypes)
```
